# **EDITING SERIES**

```python
marks = [97,99,78,65,100]
subjects = ['maths','english','science','CS','DS']

# Series in such a way that marks --> value and subjects --> Index
pd.Series(marks,index = subjects)

marks = pd.Series(marks,index = subjects,name = 'My marks')
marks

marks = {
    'maths': 97,
    'english' : 99,
    'science' : 78,
    'CS' : 65,
    'DS' : 100
}

marks_series = pd.Series(marks,name = 'Mymarks')
marks_series

# Dictionary pass karne se -- automatically keys and index set karlete h !!
```

1. Using Indexing

```python
marks_series.iloc[1] = 100 
marks_series
# if we have integer index we sue iloc ie integer location for locating index !!!
```

```python
"""What if and index doesnt exist ? """

marks_series['Social Studeis'] = 50
marks_series

# New item is added in the series if we do marks_series['evs']--> this will throw an error coz this is readung ones 
```

2. Slicing

```python
runs = [13,24,56,78,100]
runs_ser = pd.Series(runs)
runs_ser
```

```python
runs_ser[2:4] = 100
runs_ser

# this edits the selected index and then make changes
# we can edit multiple items in a single go !!
```

3. Fancy Indexing

```python
# if we want to put zerso in 0.1 and 4 index 

runs_ser[[0,3,4]] = [0,0,0]
runs_ser
```

4. Using Index label

```python
bollywood['2 States (2014 film)'] = 'Alia Bhatt'
bollywood['2 States (2014 film)']

# in this we used index as label for editing Series 
```
