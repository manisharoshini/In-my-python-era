### 1. Pandas Series

A pandas series is like a column in a table. It is a 1D Array holding data of any type.

Importing Pandas

```python
import numpy as np # pandas is made above numpy.. so it is a good practice to import numpy as well
import pandas as pd
```

#### Series from List

1. String

```python
country = ['India','Pakistan','USA','South Korea','Nepal']
pd.Series(country)
```

2. Integer

```python
runs = [12,24,56,78,100]
pd.Series(runs)
```

In all above series the indes is generated automatically... here we can also custom index

3. Custom a Index

```python
marks = [97,99,78,65,100]
subjects = ['maths','english','science','CS','DS']

# Series in such a way that marks --> value and subjects --> Index
pd.Series(marks,index = subjects)
```

4. Setting a name to Series

```python
marks = pd.Series(marks,index = subjects,name = 'My marks')
marks
```

#### Series from Dict

```python
marks = {
    'maths': 97,
    'english' : 99,
    'science' : 78,
    'CS' : 65,
    'DS' : 100
}

marks_series = pd.Series(marks,name = 'Mymarks')
marks_series

# Dictionary pass karne se -- automatically keys and index set karlete h !!
```
