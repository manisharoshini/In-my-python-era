The series is created with existig dataset ie. Real world datasets

Read_csv - csv is an excel like data known as comma seperated files. this function extracts data from csv files.

```python
import pandas as pd
```

Dataset with one column

```python
subs = pd.read_csv('Datasets/subs.csv')
subs
```

```python
type(subs) # the dataset by defaulty stores data in the form of the DataFrame not Series. we have to explicitly convert datasets into dataframe
```

```python
# To explicitly convert datasets in the form of Series 
subs = subs.squeeze() # squeeze() - converts dataFrame into Series
print(type(subs))
```

5. sort_values --> inplace

```python
ipl.sort_values()
```

```python
ipl.sort_values(ascending=False)
```

```python
# if we want to print inlt top row 
ipl.sort_values(ascending=False).head(1)
```

```python
# if we want to prient only number:
ipl.sort_values(ascending=False).head(1).values
```

```python
# if we want only the number not the array:

ipl.sort_values(ascending=False).head(1).values[0] 
# --> this is known as method chaining -- The output of first method becomes the input of next method  and the output then becomes input of next methods4

"""this only did temproary changes in the series -- if we wanto change the original series -- we use ipl.sort_values(inplace = True)"""
```
