15. Write a Python code to concatenate two NumPy arrays horizontally.

```python
h = np.arange(1,10).reshape(3,3)
i = np.arange(11,20).reshape(3,3)

print("Concatination using concatenate() function: ")
print(np.concatenate((h,i),axis = 1))
print("The horizontal concatenation of h and i is:")
print(np.hstack((h,i)))
```

16. Write a Python code to concatenate two NumPy arrays vertically.

```python
h = np.arange(1,10).reshape(3,3)
i = np.arange(11,20).reshape(3,3)

print("Concatination using concatenate() function: ")
print(np.concatenate((h,i),axis = 0))
print("The vertical concatenation of h and i is:")
print(np.vstack((h,i)))
```
