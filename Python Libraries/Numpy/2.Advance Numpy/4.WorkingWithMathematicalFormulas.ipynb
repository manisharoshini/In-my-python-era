***WORKING WITH MATHEMATICAL FORMULAS***

```python
import numpy as np
```

**Built-in functions**

```python
a = np.arange(10)
a
```

```python
# Add all the elements of array a
print(np.sum(a))

# calculate sine value:
print(np.sin(a))
```

These are some buildin functions that we use commonly so numpy have made it as a builtin functions. So that we can work faster

**How to create functions which are not in Builtin numpy**

```python
# sigmoid (here we are creating our own functions):
def sigmoid(arr):
    return 1/(1+np.exp(-(arr)))

print(sigmoid(a))
```

```python
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)

print(actual)
print(predicted)
```

```python
# Mean Square error 

def MSE(actual,predicted):
    return np.mean((actual-predicted)**2) # 1/n --> means 'mean' value. or else we can use
# n = len(actual) 
# return (1 / n) * np.sum((actual - predicted) ** 2)

print(MSE(actual,predicted))
```
