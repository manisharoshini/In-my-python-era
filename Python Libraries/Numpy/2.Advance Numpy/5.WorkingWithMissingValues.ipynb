***WORKING WITH MISSING VALUES***

```python
import numpy as np
```

working with missing values --> np.nan

```python
arr = np.array([1,2,3,4,np.nan,6])
arr
```

```python
# boolean indexing:

# Step1: Check whetehr there's any missing or nan values in the array or not
np.isnan(arr)
```

```python
# Step2: Take only values:
arr[~np.isnan(arr)] # --> This will remove all the missing values ie. nan values in the array arr '~' -> invert 
```
