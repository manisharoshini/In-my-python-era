BROADCASTING

```python
import numpy as np
```

```python
# same shape 
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print(a)
print(b)

print(a+b)
```

```python
# different shape:

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a)
print(b)
print("----------------------------------")
print(a+b)
```
