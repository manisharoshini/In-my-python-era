BROADCASTING

```python
import numpy as np
```

```python
# same shape 
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)
print(a)
print(b)

print(a+b)
```

```python
# different shape:

a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
print(a)
print(b)
print("----------------------------------")
print(a+b)
```

Examples based on Broadcasting: 

```python
a = np.arange(12).reshape(4,3)
b = np.arange(3)

print(a)
print(b)
print("Addition of a and b is: ")
c = a + b
print(c)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)
print(a+b)

# this will throw an error like (3,4) and (,3) cannot be broadcasted
# (,3) --> (1,3) --> (3,3) ---> (3,4) and (4,3) is not equal so it throws an error
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)
print(a)
print()
print(b)
print("Addition of a and b: ")
print(a+b)

# array a(1,3) --> (3,3) and array b(3,1) --> (3,3) here both the arrays are stretched 
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)

print(a)
print()
print(b)
print("Addition of a and b: ")
print(a+b)


# array a(1,3) --> (4,3) 
# array b(4,1) --> (4,3)
```

```python
a = np.array([1])
b = np.arange(4).reshape(2,2)
print(a)
print()
print(b)
print("Addition of a and b:")
print(a+b)

# here array a(1,1) and array b(2,2) after streching the element of size 1 we get --> a(2,2) and b(2,2)
```
