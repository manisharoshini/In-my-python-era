ADAVNCED INDEXING

Indexing - we want to pick specific element from numpy array

```python
import numpy as np
```

```python
a = np.arange(24).reshape(6,4)
a
```

There are two types of indexing:
1. Fancy Indexing
2. Boolean Indexing

1. Fancy Indexing

```python
# if we want to print 1,3,and 4th row:
a[[0,2,3]]
```

```python
# 1st row, 3rd row, 4th row and last row:
a[[0,2,3,5]]
```

```python
# if we want to print 1st column 2nd column and 4th column:
a[:,[0,1,3]] # : -> all rows, [0,1,3] -> means the column we want to select
```
