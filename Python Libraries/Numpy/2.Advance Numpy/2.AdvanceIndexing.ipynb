ADAVNCED INDEXING

Indexing - we want to pick specific element from numpy array

```python
import numpy as np
```

```python
a = np.arange(24).reshape(6,4)
a
```

There are two types of indexing:
1. Fancy Indexing
2. Boolean Indexing

1. Fancy Indexing

```python
# if we want to print 1,3,and 4th row:
a[[0,2,3]]
```

```python
# 1st row, 3rd row, 4th row and last row:
a[[0,2,3,5]]
```

```python
# if we want to print 1st column 2nd column and 4th column:
a[:,[0,1,3]] # : -> all rows, [0,1,3] -> means the column we want to select
```

2. Boolean Indexing

```python
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
# Find all the numbers greater than 50:
a>50 # this will give an boolean array
```

```python
a[a>50]
# This is used to filter data based on certain condition 
```

```python
# FInd all even numbers in array a
a[a%2 == 0]
```

```python
# find all numbers that are greater than 50 and are evn:
a[(a>50)&(a%2 == 0)]
```

```python
# Find all numbers Divisible by 7:
a[a%7 == 0]
```

```python
# find all numbers that are not divisbe by 7:
print(a[a%7 != 0]) # my logic
print(a[~(a%7 == 0)]) # given logic
```
