Numpy Array vs Python Arrays 

```python
import numpy as np
```

1. Based on Speed

```python
# Using Lists: 

a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

import time

Start = time.time()

c = []
for i in range(len(a)):
    c.append(a[i] + b[i])


print(time.time() - Start)

# the time took to execute this code is "1.4742190837860107" (in Youtube its 3mins)
```

```python
# Using Numpy:

a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = time.time()
c = a + b
print(time.time() - Start)

# the time only took 0.3 or 0.1 secs
```

Numpy is 50x faster thahn python lists

Numpy array uses Static C type array ie it is a static array and size is fixed.. 
whereas the python List is a dynamic array the size is doubled when we add extra elements 

Numpy array is not a referential array, means we can store away elements directly in memory..
whereas python lists uses referntial memory ie. it stores only the memory location. 
Then python gets memory location and manually search for items

2. Based on Memory

```python
import sys
a = [i for i in range(10000000)]
sys.getsizeof(a) #  -- this tells how many bytes it occupies 
```

```python
arr = np.arange(10000000,dtype = np.int32) # we can also use int16 or int8 for more reduction of space 
sys.getsizeof(arr)

# int 64 almost uses same amount of space.. but here we have freedom to reduce space... we used int32 to reduce the unwanted space. 
# This is possible in numpy
```
