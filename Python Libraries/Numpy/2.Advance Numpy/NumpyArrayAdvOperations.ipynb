Numpy Array vs Python Arrays 

```python
import numpy as np
```

```python
# Using Lists: 

a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

import time

Start = time.time()

c = []
for i in range(len(a)):
    c.append(a[i] + b[i])


print(time.time() - Start)

# the time took to execute this code is "1.4742190837860107" (in Youtube its 3mins)
```

```python
# Using Numpy:

a = np.arange(10000000)
b = np.arange(10000000, 20000000)

start = time.time()
c = a + b
print(time.time() - Start)

# the time only took 0.3 or 0.1 secs
```
