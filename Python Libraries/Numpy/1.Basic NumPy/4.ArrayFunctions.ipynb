ARRAY FUNCTIONS

```python
import numpy as np
```

```python
# -- numpy has a lot of build in functions for every mathematical operations 
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
print(a1)
```

1. Basic Functions(Max,Min,Sum,Prod)

```python
print(f"The maximum of a1 is {np.max(a1)}")
print("---------------------------------")
print(f"The Minimum of a1 is {np.min(a1)}")
print("---------------------------------")
print(f"The Sum of a1 is {np.sum(a1)}")
print("---------------------------------")
print(f"The Product of the matrix is {np.prod(a1)}")
```

```python
# we can access the max,min,sum,prod of each row or coloumn
# 0 -> col and 1 -> row
print(f"The maximum of each row is {np.max(a1,axis = 1)}")
print(f"The maximum of each column is {np.max(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The minimum of each row is {np.min(a1,axis = 1)}")
print(f"The minimum of each column is {np.min(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The product of each row is {np.prod(a1,axis = 1)}")
print(f"The product of each column is {np.prod(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The Sum of each row is {np.sum(a1,axis = 1)}")
print(f"The Sum of each column is {np.sum(a1,axis = 0)}")
```

2. Statstical Functions (mean,median,std,var)

```python
print(f"The mean of the array a1 is {np.mean(a1)}")
print("-------------------------------------------------------------------")
print(f"The median of array a1 is {np.median(a1)}")
print("-------------------------------------------------------------------")
print(f"The Standard Deviations of array a1 is {np.std(a1)}")
print("-------------------------------------------------------------------")
print(f"The Variation of array a1 is {np.var(a1)}")
```

```python
# -- here we have a freedom to select we want row or column
# 0 -> col and 1 -> row
# Mean --> calculates the average 
# Median --> sorts the entire thing array and gives the median values
print(f"The Mean of a1 row-wise is {np.mean(a1,axis = 1)}")
print(f"The Mean of a1 column-wise is {np.mean(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The Median of a1 row-wise is {np.median(a1,axis = 1)}")
print(f"The Median of a1 column-wise is {np.median(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The Standard Deviation of a1 row-wise is {np.std(a1,axis = 1)}")
print(f"The Standard Deviation of a1 column-wise is {np.std(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The Variance of a1 row-wise is {np.var(a1,axis = 1)}")
print(f"The Variance of a1 column-wise is {np.var(a1,axis = 0)}")
```

3. Trignometric Functions (sin,cos,tan)

```python
print(f"The sine of array a1 is {np.sin(a1)}")
print("-------------------------------------------------------------------")
print(f"The cos of array a1 is {np.cos(a1)}")
print("-------------------------------------------------------------------")
print(f"The tan value of array a1 is {np.tan(a1)}")
```

DOT PRODUCT

```python
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)
print(a2)
print("-------------------------------------------------------------------")
print(a3)
```

```python
print("The dot product if two matrix is: ")
print(np.dot(a2,a3)) # -- This will be an (3x3) matrix 
```

Log and Exponents

```python
print("The log of each item in a1 matrix is: ")
print(np.log(a1))
print("-------------------------------------------------------------------")
print("The Exponents of each value in a1 matrix are: ")
print(np.exp(a1))
```

Round, Floor and Ceil

```python
a4 = np.random.random((2,3))*100
print(a4)
```

```python
print(np.round(a4)) # -- round function will give the integer values
```

```python
print(np.floor(a4)) # -- this will give the exact value and ignores everything after pointer even its greater than .5
# eg: the value generated is 6.98..
# --- round will give output as 7 
# --- floor will give output as 6
```

```python
print(np.ceil(a4)) # -- this will give the front value of decimal (never ignores anything after decimal)
# eg: the value generated is 6.12..
# --- round will give output as 6
# --- floor will give output as 6
# --- ceil will give output as 7 
```
