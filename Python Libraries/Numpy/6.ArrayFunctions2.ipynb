Iterating

```python
import numpy as np

a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
print(a1)
print("--------------------------------------------------")
print(a2)
print("--------------------------------------------------")
print(a3)
```

1. 1D Array

```python
for i in a1:
    print(i)
```

2.2D Array

```python
# this fetches a whole row one at a time
for i in a2:
    print(i)
print("---------------------------------")
# if we want to fetch single item from array:
for i in np.nditer(a2):
    print(i)
```

2. 3D Array

```python
# below code fetches the whole matrix one at a time:
for i in a3:
    print(i)
print("---------------------------------")
# if we want to fetch single element:
for i in np.nditer(a3):
    print(i)
```

```python
print(np.transpose(arr))
print("----------------------------")
print(a2.T) # this also does same function
```
