Array Slicing and Indexing

```python
import numpy as np
```

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
print(a1)
print("--------------------------------------------------")
print(a2)
print("--------------------------------------------------")
print(a3)
```

--Indexing--

1. 1D array

```python
# If we want a last element in Python: 
print(a1[-1])
```

```python
# If we want 1st element:
print(a1[0])
```

2. 2D Array

```python
a2
```

```python
# print the value '6' from the matrix: 
# [row,column], Row - Horizontal and column - vertical 

print(a2[1,2])
```

```python
# Print '11' from the matrix:
print(a2[2,3])
```

```python
# Print '4' from a2:
print(a2[1,0])
```

3. 3D Array

```python
a3
```

```python
# print '5' from a3 [3d Matrix]:
# 1 -> indicates the in ehich matrix the element is located
# 0 -> row (same as 2d matrix)
# 1 -> column (same as 2d matrix)


print(a3[1,0,1])
```

```python
# print '2' from a3 matrix:
print(a3[0,1,0])
```

```python
# print '0 ' from a3: 
print(a3[0,0,0])
```

```python
# print '6' from a3:
print(a3[1,1,0])
```

--Slicing--

1. 1D Array

```python
a1
```

```python
# print 2,3,4 -- (This is known as slicing)

print(a1[2:5])
```

```python
# Start : end : strides(steps)

print(a1[2:5:2]) # thus will print alternatives
```
