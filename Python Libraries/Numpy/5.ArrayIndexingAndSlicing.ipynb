Array Slicing and Indexing

```python
import numpy as np
```

```python
a1 = np.arange(10)
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
print(a1)
print("--------------------------------------------------")
print(a2)
print("--------------------------------------------------")
print(a3)
```

--Slicing--

1. 1D Array

```python
a1
```

```python
# print 2,3,4 -- (This is known as slicing)

print(a1[2:5])
```

```python
# Start : end : strides(steps)

print(a1[2:5:2]) # thus will print alternatives
```

2. 2D Array

```python
a2
```

```python
# print only 1st row ie 0,1,2,3 from a2: 
print(a2[0,:]) # : all elements 
```

```python
# print only 3rd column (2,6,10):
print(a2[:,2] ) 

# : --> all elements in the row
# 2 --> represents the column we want 
```

```python
# if we ant 5,6 and 7,9: (Square sub-matrix):
print(a2[1:,1:3])
```

```python
# we want 0,38,11:
a2[::2,::3]
```

```python
# print 1,3,9,11:
a2[::2,1::2]
```

```python
# print only 4,7:
print(a2[1:2,0::3])   # my logic 
print(a2[1,::3]) # sir's logic 
```

```python
a2
```

```python
# print 1,2,3,5,6,7
print(a2[:2,1:4])
```

```python
# print 1,3,5,7
print(a2[:2,1::2])
```
