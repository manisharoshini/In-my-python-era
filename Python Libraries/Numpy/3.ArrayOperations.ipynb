ARRAY OPERATIONS

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
print(a1)
print("---------------------------------")
print(a2)
```

1. Scalar Operations: 

```python
print("Multiplication: ")
print(a1*2)
print("---------------------------------")

print("Addition: ")
print(a1+2)
print("---------------------------------")

print("Subtraction: ")
print(a1-2)
print("---------------------------------")

print("Division: ")
print(a1/2)
print("---------------------------------")

print("Modulus:")
print(a1%2)
print("---------------------------------")

print("Exponential: ")
print(a1**2)
print("---------------------------------")
```

2. Relational Operators

```python

print("checking if each and every elements in the matrix is greater than 5:")
print(a2>5)
print("---------------------------------")

print(a2>15)
print("---------------------------------")

print(a2 == 15)
print("---------------------------------")
```
