ARRAY OPERATIONS

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)
print(a1)
print("---------------------------------")
print(a2)
```

1. Scalar Operations: 

```python
print("Multiplication: ")
print(a1*2)
print("---------------------------------")

print("Addition: ")
print(a1+2)
print("---------------------------------")

print("Subtraction: ")
print(a1-2)
print("---------------------------------")

print("Division: ")
print(a1/2)
print("---------------------------------")

print("Modulus:")
print(a1%2)
print("---------------------------------")

print("Exponential: ")
print(a1**2)
print("---------------------------------")
```

2. Relational Operators

```python

print("checking if each and every elements in the matrix is greater than 5:")
print(a2>5)
print("---------------------------------")

print(a2>15)
print("---------------------------------")

print(a2 == 15)
print("---------------------------------")
```

3. Vector Operations

```python
print(a1)
print("---------------------------------")
print(a2)
print("---------------------------------")

# Note: -- This opertaions only works when the shapes are same 
print("Addition of Two matrices: ")
print(a1 + a2)

print("---------------------------------")

print("Subtarction of Two matrices: ")
print(a2 - a1)

print("---------------------------------")

print("Multiplication of two matrices: ")
print(a1 * a2)

print("---------------------------------")

print("Division of Two Matrices: ")
print(a1/a2)

print("---------------------------------")

print("Power/ Exponential of two Matrices: ")
print(a1**a2)
```

ARRAY FUNCTIONS

```python
# -- numpy has a lot of build in functions for every mathematical operations 
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
print(a1)
```

1. Max,Min,Sum,Prod

```python
print(f"The maximum of a1 is {np.max(a1)}")
print("---------------------------------")
print(f"The Minimum of a1 is {np.min(a1)}")
print("---------------------------------")
print(f"The Sum of a1 is {np.sum(a1)}")
print("---------------------------------")
print(f"The Product of the matrix is {np.prod(a1)}")
```

```python
# we can access the max,min,sum,prod of each row or coloumn
# 0 -> col and 1 -> row
print(f"The maximum of each row is {np.max(a1,axis = 1)}")
print(f"The maximum of each column is {np.max(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The minimum of each row is {np.min(a1,axis = 1)}")
print(f"The minimum of each column is {np.min(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The product of each row is {np.prod(a1,axis = 1)}")
print(f"The product of each column is {np.prod(a1,axis = 0)}")
print("-------------------------------------------------------------------")
print(f"The Sum of each row is {np.sum(a1,axis = 1)}")
print(f"The Sum of each column is {np.sum(a1,axis = 0)}")


```
